```python
top_10 = df.sort_values(by='Rank', ascending=True).head(10)
top_10[['Language', 'Year_Introduced', 'Job_Demand_Index', 'Rank']]
```

```python
plt.figure(figsize=(10, 5))

# Gráfico de barras: Eje X = Lenguaje, Eje Y = Demanda Laboral
plt.bar(top_10['Language'], top_10['Job_Demand_Index'], color='skyblue')

plt.title('Demanda Laboral de los 10 Lenguajes Mejor Rankeados')
plt.xlabel('Lenguaje de Programación')
plt.ylabel('Índice de Demanda Laboral (0 - 100)')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(10, 6))

# Cambiamos 'Rating_Pct' por 'Job_Demand_Index'
sns.histplot(df['Job_Demand_Index'], bins=10, kde=True, color='teal')

plt.title("Distribución del Índice de Demanda Laboral")
plt.xlabel("Índice de Demanda Laboral (0 - 100)")
plt.ylabel("Cantidad de Lenguajes")
plt.show()
```

```python
df_modernos = df[(df['Year_Introduced'] >= 2000) & (df['Year_Introduced'] <= 2021)]

top_modernos = df_modernos.sort_values(by='Job_Demand_Index', ascending=False).head(10)

plt.figure(figsize=(10, 5))

sns.barplot(
    data=top_modernos,
    x='Language',
    y='Job_Demand_Index',
    palette='magma'
)

plt.title("Demanda Laboral de Lenguajes Modernos (Creados entre 2000 y 2021)")
plt.xlabel("Lenguaje")
plt.ylabel("Índice de Demanda Laboral (0 - 100)")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
df = df.drop(columns=['Creator'])

# Guardamos el archivo corregido
df.to_csv('PopularidadLenguajes.csv', index=False)
print("Archivo corregido con éxito")
```
